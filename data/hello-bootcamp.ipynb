```python
# Check: which versions do you have?
import sys
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt

print("Python    ", sys.version.split()[0])
print("NumPy     ", np.__version__)
print("pandas    ", pd.__version__)
print("matplotlib", matplotlib.__version__)
```

```python
# Load: penguins.csv must be in the same folder as this notebook
import os

try:
    df = pd.read_csv("penguins.csv")
except FileNotFoundError:
    print("I cannot find penguins.csv in this folder:")
    print(os.getcwd())
    print("Save penguins.csv in that folder (next to this notebook), then run this cell again.")
    raise

print(df.shape[0], "penguins,", df.shape[1], "columns")
df.head()
```

```python
# Plot: one dot per penguin, with labelled axes
fig, ax = plt.subplots()
ax.scatter(df["flipper_length_mm"], df["body_mass_g"])
ax.set_xlabel("Flipper length (mm)")
ax.set_ylabel("Body mass (g)")
ax.set_title("Palmer penguins")
plt.show()
```
